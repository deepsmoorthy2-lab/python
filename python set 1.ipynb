```python
def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    return fact

print(factorial(5))
```

```python
num=int(input("Enter a number: "))
count=0
for i in range(1,num+1):
    if num%i==0:
        count=count+1
if count==2:
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```
